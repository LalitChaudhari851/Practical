```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
```

```python
# 1. Load the dataset
df = pd.read_csv('Bengaluru_House_Data.csv')

# 2. Explore the dataset
print("DataFrame Head:")
print(df.head())

print("\nDataset Shape:", df.shape)
print("\nDataset Information:")
df.info()
```

```python
# 3. Handle missing values
print("\nMissing Values Before Handling:")
print(df.isnull().sum())

df['balcony'] = df['balcony'].fillna(df['balcony'].median())
df['bath'] = df['bath'].fillna(df['bath'].median())
df['society'] = df['society'].fillna('Unknown')

df.dropna(subset=['location', 'size'], inplace=True)

print("\nMissing Values After Handling:")
print(df.isnull().sum())
```

```python
# 4. Create and merge DataFrames
neighborhood_info = {
    'location': ['Electronic City Phase II', 'Chikka Tirupathi',
                 'Uttarahalli', 'Whitefield'],
    'neighborhood_quality': ['Good', 'Average', 'Good', 'Excellent']
}

df_neighborhood = pd.DataFrame(neighborhood_info)

df_merged = pd.merge(
    df, df_neighborhood, on='location', how='left'
)

print("\nMerged DataFrame:")
print(df_merged.head())
```

```python
# 5. Filtering
ready_to_move = df[df['availability'] == 'Ready To Move']
print("\nReady To Move:", ready_to_move.shape)

two_bhk = df[df['size'] == '2 BHK']
print("2 BHK Properties:", two_bhk.shape)

whitefield = df[df['location'] == 'Whitefield']
print("Whitefield Properties:", whitefield.shape)

specific = df[
    (df['availability'] == 'Ready To Move') &
    (df['size'] == '2 BHK') &
    (df['location'] == 'Whitefield')
]

print("\nReady To Move 2 BHK in Whitefield:")
print(specific.shape)
print(specific.head())
```

```python
# 6. One-hot encoding of area_type
area_dummies = pd.get_dummies(df['area_type'], prefix='area')
df_encoded = pd.concat([df, area_dummies], axis=1)

print("\nOne-Hot Encoded Data:")
print(df_encoded.head())
```

```python
# 7. Label encoding of location
label_encoder = LabelEncoder()
df['location_encoded'] = label_encoder.fit_transform(df['location'])

print("\nLabel Encoded Location:")
print(df[['location', 'location_encoded']].head())
```

```python
# 8. Calculate average prices by location
avg_price_location = (
    df.groupby('location')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Price by Location (Top 10):")
print(avg_price_location.head(10))

# 9. Calculate average prices by area type
avg_price_area = (
    df.groupby('area_type')['price']
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage Price by Area Type:")
print(avg_price_area)
```

```python
# 10. Handle outliers
# Convert total_sqft ranges into a single number
def convert_sqft_to_num(x):
    try:
        tokens = str(x).split('-')

        if len(tokens) == 2:
            return (float(tokens[0]) + float(tokens[1])) / 2

        return float(x)

    except (ValueError, TypeError):
        return np.nan

df['total_sqft'] = df['total_sqft'].apply(convert_sqft_to_num)

# Remove rows where total_sqft is invalid
df.dropna(subset=['total_sqft'], inplace=True)

# Calculate IQR for price
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\nOriginal Shape:", df.shape)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

# Remove price outliers
df_no_outliers = df[
    (df['price'] >= lower_bound) &
    (df['price'] <= upper_bound)
]

print("\nShape After Removing Outliers:")
print(df_no_outliers.shape)
```

```python
# 11. Export cleaned dataset
df_no_outliers.to_csv('Bengaluru_House_Data_Cleaned.csv', index=False)

print("\nCleaned dataset exported successfully!")
```
