```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 1: Load data from different files
csv_df = pd.read_csv('sales_data_sample.csv', encoding='latin1')
excel_df = pd.read_excel('sales_data_sample.xlsx')
json_df = pd.read_json('sales_data_sample.json')

print("CSV, Excel and JSON files loaded successfully")
```

```python
# Step 2: Combine and clean data
sales_df = pd.concat([csv_df, excel_df], ignore_index=True)

print("Before cleaning:", sales_df.shape)

sales_df.drop_duplicates(inplace=True)
sales_df.fillna(0, inplace=True)

print("After cleaning:", sales_df.shape)
print(sales_df.head())
```

```python
sales_df = pd.concat([csv_df,excel_df],ignore_index='False')

print("Before Cleaning : ",sales_df.shape)

sales_df.drop_duplicates(inplace=True)
sales_df.fillna(0,inplace=True)

print("After Cleaning : ",sales_df.shape)
print(sales_df.head())
```

```python
# Step 3: Transform data
sales_df['TotalSales'] = (
    sales_df['QUANTITYORDERED'] * sales_df['PRICEEACH']
)

sales_df['ORDERDATE'] = pd.to_datetime(
    sales_df['ORDERDATE'], errors='coerce'
)

print(sales_df.head())
```

```python
# Step 4: Analyze data
print("\nDescriptive Statistics:")
print(sales_df[['QUANTITYORDERED', 'PRICEEACH', 'TotalSales']].describe())

print("Total Revenue:", sales_df['TotalSales'].sum())
```

```python
# Step 5: Visualize data
sns.set_theme(style="whitegrid")

# 1. Sales by Year
if 'YEAR_ID' in sales_df.columns:
    sales_df.groupby('YEAR_ID')['TotalSales'].sum().plot(kind='bar')
    plt.title("Total Sales by Year")
    plt.xlabel("Year")
    plt.ylabel("Total Sales")
    plt.tight_layout()
    plt.show()

```

```python
# 2. Monthly Sales Trend
monthly_sales = sales_df.set_index('ORDERDATE')['TotalSales'].resample('ME').sum()
monthly_sales.plot(kind='line', marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()

```

```python
# 3. Order Status Distribution
sns.countplot(data=sales_df, x='STATUS')
plt.title("Order Status Distribution")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


```

```python
# 4. Sales Distribution
sns.histplot(data=sales_df, x='TotalSales', bins=50, kde=True)
plt.title("Sales Amount Distribution")
plt.xlabel("Total Sales")
plt.tight_layout()
plt.show()
```
