# IR Practical 1: Text Preprocessing

```python
import nltk
import re
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer,WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("averaged_perceptron_tagger_eng")
nltk.download("omw-1.4")
```

```python

text = """The students are studying different subjects.
They are learning programming and playing games."""

# Convert to lowercase
text = text.lower()

# Remove punctuation
text = re.sub(r'[^a-zA-Z\s]', '', text)

# 1. Tokenisation
tokens = word_tokenize(text)
print("Tokens:", tokens)
```

```python

stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in tokens if word not in stop_words
]

print("After Stop Word Removal:", filtered_words)
```

```python

stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word) for word in filtered_words
]

print("After Stemming:", stemmed_words)
```

```python

lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word) for word in filtered_words
]

print("After Lemmatization:", lemmatized_words)
```
